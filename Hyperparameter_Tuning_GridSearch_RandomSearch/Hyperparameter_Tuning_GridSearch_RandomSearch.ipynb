# HYPER PARAMETER TUNNING

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('iris')
df.head()
```

```python
df['species'].unique()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x = df.drop('species', axis=1)
y = df['species']
```

```python
x.head()
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

KNN

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=5) 
```

```python
knn.fit(x_train, y_train)
```

```python
knn.score(x_test, y_test)
```

SUPPORT VECTOR MACHINE

```python
from sklearn.svm import SVC
svm = SVC()
```

```python
svm.fit(x_train, y_train)
```

```python
svm.score(x_test, y_test)
```

GRID SEARCH CV

```python
from sklearn.model_selection import GridSearchCV
```

GRIDSEARCH_CV THROUGH KNN

```python
classifiers2 = GridSearchCV(knn,{
    'n_neighbors' : [3, 5, 7, 9],
    'weights' : ['uniform', 'distance'],
    'algorithm' : ['auto', 'ball_tree', 'kd_tree', 'brute']
}, cv = 5,
return_train_score = True)
```

```python
classifiers2.fit(x, y)
```

```python
classifiers2.cv_results_
```

```python
results_df2 = pd.DataFrame(classifiers2.cv_results_)
results_df2.head()
```

training scores

```python
print(classifiers2.cv_results_['mean_train_score'])
```

validation scores

```python
print(classifiers2.cv_results_['mean_test_score'])
```

GRIDSEARCH_CV THROUGH SVM

```python
classifiers = GridSearchCV(svm, {
    'kernel': ['linear', 'rbf'],
    'C': [0.1, 1, 10],
    'gamma': ['scale', 'auto']
}, cv=5, return_train_score = False)
```

```python
classifiers.fit(x, y)
```

```python
classifiers.cv_results_
```

```python
results_df = pd.DataFrame(classifiers.cv_results_, columns=['mean_test_score', 'std_test_score', 'params'])
results_df
```

RANDOM SEARCH CV

#it is used for big data models \n
#Although this data is small but we are using it just for practice

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
rand_classifier = RandomizedSearchCV((svm), {
    'kernel': ['linear', 'rbf'],
    'C': [0.1, 1, 10],
    'gamma': ['scale', 'auto']
   }, n_iter=4, cv=5, return_train_score = False)
```

```python
rand_classifier.fit(x, y)
```

```python
rand_result = pd.DataFrame(rand_classifier.cv_results_, columns=['param_C', 'param_kernel', 'mean_test_score'])
rand_result
```
